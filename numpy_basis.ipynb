Function of NP Array


1.ARANGE

```python
arr_1 = np.arange(10,100,5)
arr_1
```

```python
arr_1 = arr_1.reshape(3,3,2)
arr_1
```

2.Linspace

```python
arr_1 = np.linspace(1,100,num=18).astype('int')
arr_1
```

```python
arr_1.reshape(3,6)
```

3.ONES

```python
arr_1 = np.ones(25).astype('int')
arr_1
```

```python
arr_1 = arr_1.reshape(5,5)
arr_1
```

```python
arr_2 = np.zeros(15).astype('int')
arr_2
```

```python
arr_2 = arr_2.reshape(3,5)
arr_2
```

```python
arr3 = np.arange(30)
arr3
```

```python
arr3.reshape(2,5,3)
```
